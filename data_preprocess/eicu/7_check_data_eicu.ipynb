```python
# Import libraries
import pandas as pd
import os
import matplotlib.pyplot as plt
import numpy as np
from pathlib import Path

from utils.seed import set_random_seed
from utils.constants import EICU_RARE_ICD_CODES
from utils.data import get_info_eicu
```

```python
timeseries_dir = r"data/eICU_first24h_ts1h/timeseries"
all_dir = r"data/eICU_first24h_ts1h/all"
ethnicities = ['Asian', 'Black', 'Hispanic', 'White', 'Unknown', "Other"]

set_random_seed(27)

demo_df = pd.read_csv(os.path.join(timeseries_dir, "demographics.csv"), sep=',')
ts_df = pd.read_csv(os.path.join(timeseries_dir, "time-series.csv"), sep=',')
label_df = pd.read_csv(os.path.join(all_dir, "label_icumortality.csv"), sep=',')
```

## Check Demographics

```python
# get icd92stay_ids
icd92stay_ids = {}
for icd9_code in EICU_RARE_ICD_CODES:
    icd92stay_ids[icd9_code] = demo_df[demo_df.icd9_code==icd9_code].stay_id.values

# print(icd92hadm_ids)
```

```python
# get info for the ICD_038 of each ethnicity
info_dict = {}

for target_icd in EICU_RARE_ICD_CODES:
    info = get_info_eicu(target_icd, icd92stay_ids, demo_df, label_df, ts_df)
    info_dict[target_icd] = info

info_df = pd.DataFrame(info_dict).T
info_df['RecordNumber'] = info_df['RecordNumber'].astype(int)
info_df = info_df.sort_values(by='RecordNumber', ascending=False)

print(info_df)
info_df.to_csv("info.csv", sep=',')
```
